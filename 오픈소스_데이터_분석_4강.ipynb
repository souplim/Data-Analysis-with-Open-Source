## 4-3 텍스트 파일 읽기 및 데이터 추출

```python
import re # 정규식 라이브러리(regular expression)

## 파일(callcenter20250301.log) 오픈 및 읽기
with open('callcenter20250301.log', 'r', encoding='utf-8') as f:
    content = f.read()

## 주민등록번호 패턴 생성
pattern = re.compile(r'(\d{6})-(\d{7})') # () - () 묶음을 두 묶음으로 가져가겠다

## 주민등록번호 마스킹
masked_content = pattern.sub(r'\1-*******', content) # 첫번째 그룹의 6자리는 그대로 쓰고 뒷자리는 마스킹

## 마스킹된 파일(callcenter20250301_masked.log) 오픈 및 쓰기
with open('callcenter20250301_masked.log', mode='w') as f:
    f.write(masked_content)

print("주민등록번호 마스킹 완료. 'callcenter20250301_masked.log.txt' 파일로 저장되었습니다.")
```

## 4-4 Open-Meteo의 무료 날씨 API를 통한 특정 지역 온도 조회

```python
import requests
import json

url = "https://api.open-meteo.com/v1/forecast"
params = {
    "latitude": "37.58638333",
    "longitude": "127.0203333",
    "current": "temperature_2m"
}

try:
    ## URL 및 파라미터 전송
    response = requests.get(url, params = params)
    response.raise_for_status() # 에러 처리 -> exception 발생하도록

    ## JSON 데이터 읽기
    data = response.json()

    print("API 응답:", data)

    # 1) 값 꺼내기: data는 API가 돌려준 딕셔너리(dict)입니다.
    #    실제 응답은 대략 이런 모양이에요:
    #    {
    #      "current_units": {"temperature_2m": "°C"},
    #      "current":       {"temperature_2m": 18.3}
    #    }
    #
    #    data['current']['temperature_2m']        -> 18.3  (온도 숫자)
    #    data['current_units']['temperature_2m']  -> "°C"  (단위 문자열)
    #    -> 딕셔너리 안의 딕셔너리라서 대괄호 []를 두 번 써서 한 단계씩 들어갑니다.

    # 2) .format(): 문자열 속의 {} 자리에 값을 순서대로 끼워 넣는 기능입니다.
    #    {0} -> format()의 첫 번째 인자 (온도 숫자)
    #    {1} -> format()의 두 번째 인자 (단위)
    print("서울시 종로구의 현재 온도는 : {0}{1} 입니다.".format(
        data['current']['temperature_2m'],        # {0} 자리에 들어감 -> 18.3
        data['current_units']['temperature_2m']   # {1} 자리에 들어감 -> °C
    ))
    # 출력 결과: 서울시 종로구의 현재 온도는 : 18.3°C 입니다.

except requests.exceptions.RequestException as e:
    print(f"API 호출 실패: {e}")
except json.JSONDecodeError as e:
    print(f"JSON 파싱 실패: {e}")
```

## 4-5 Selenium과 lxml을 이용한 웹 스크래핑

```python
!curl -o google-chrome-stable_current_amd64.deb https://dl.google.com/linux/direct/google-chrome-stable_current_amd64.deb # 웹 컨텐츠 가져오기 위한 웹 브라우저 사용하기 위해 설치 파일 다운로드
!apt install ./google-chrome-stable_current_amd64.deb -y # 설치
!pip install selenium webdriver_manager # 셀레니움 사용
```

```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service as ChromeService
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.common.by import By
from lxml import html
import time

chrome_options = webdriver.ChromeOptions()
chrome_options.add_argument('--headless')               # 브라우저 창 없이 실행!!!!!!!
chrome_options.add_argument('--no-sandbox')             # 보안모드 비활성화 (Colab 필수)
chrome_options.add_argument('--disable-dev-shm-usage')  # 메모리 부족 방지 (Colab 필수)
chrome_options.add_argument('--window-size=1920x1080')  # 창 크기 설정(가상)
chrome_options.add_argument('--disable-gpu')            # GPU 가속 비활성화 (일부 환경 안정성)
chrome_options.binary_location = "/usr/bin/google-chrome-stable"  # Colab용 크롬 경로 지정

## 드라이버 실행
driver = webdriver.Chrome(options=chrome_options)

## 사이트 접속
url = 'https://professor.knou.ac.kr/jaehwachung/index.do'
driver.get(url)

## 사이트 접속 대기
time.sleep(2) # 페이지 로드되는 시간 기다려줌

## 페이지 제목 출력
page_source = driver.page_source
tree = html.fromstring(page_source) # 보이는 사이트를 다시 html 파일로 받음 -> tree 형태로

title_text = tree.xpath('//title/text()')
print(title_text)

## 드라이버 종료
driver.quit()
```
